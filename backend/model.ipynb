```python

import pyttsx3
import time


engine = pyttsx3.init()

last_spoken = ""

while True:   # agar tum camera loop use kar rahi ho
    results = model(frame)   # ya image ka result

    labels = []

    for r in results:
        for box in r.boxes:
            cls = int(box.cls[0])
            label = model.names[cls]
            labels.append(label)

    # duplicates remove
    labels = list(set(labels))

    if labels:
        sentence = ", ".join(labels)

        # same cheez repeat na ho
        if sentence != last_spoken:
            engine.say(sentence)
            engine.runAndWait()
            last_spoken = sentence

            time.sleep(2)   # 2 sec pause (continuous bolna ruk jayega)
```

```python
import cv2
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    results = model(frame)

    annotated = results[0].plot()
    cv2.imshow("Detection", annotated)

    if cv2.waitKey(1) == 27:
        break

cap.release()
cv2.destroyAllWindows()
```
